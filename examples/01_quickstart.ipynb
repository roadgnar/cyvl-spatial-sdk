# Cyvl SDK — Quickstart

Load the Somerville scene, explore the data layers, and look at a real street photo.

```bash
pip install "cyvl[viz] @ git+https://github.com/roadgnar/cyvl-spatial-sdk"
```
The first `load_scene` call downloads ~120 MB of parquet (cached in `~/.cache/cyvl`).

````python
import cyvl

scene = cyvl.load_scene("somerville")
scene
````

Every layer is a GeoDataFrame — pavement condition, signs, distresses, markings, assets:

````python
scene.pavements[["address_st", "label", "score", "length_ft"]].sort_values("score").head(10)
````

````python
# 311k posed camera frames: every row knows exactly where the camera was and
# which way it pointed, with production-calibrated intrinsics.
scene.frames[["frame_id", "facing", "lon", "lat", "has_public_image"]].head()
````

Find photos near Davis Square and display one:

````python
near = scene.frames_near(-71.1218, 42.3967, radius_m=40, facing="front")
frame = scene.frame(near.index[0])
frame.show()
frame
````

````python
# Interactive city map of condition scores + signs (folium)
scene.map(layers=("rollup", "signs"))
````
