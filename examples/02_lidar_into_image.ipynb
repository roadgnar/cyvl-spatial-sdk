# LiDAR ↔ Imagery: the superpower

Every photo has a full 6-DoF pose in the same coordinate frame as the LiDAR.
`frame.points_in_view()` fetches the surrounding point cloud and projects it
into the photo with the exact production camera model — every 3D point gets a
pixel coordinate, every pixel can be lifted back to 3D.

```python
import cyvl

scene = cyvl.load_scene("somerville")
frame = scene.nearest_frame(-71.1218, 42.3967)
pts = frame.points_in_view(radius_m=25)
print(f"{pts.in_view.sum():,} lidar points visible in this photo")
frame.show(overlay=pts)
```

`Projected` keeps everything aligned: `.pixels`, `.depth`, `.points_utm`, `.colors`. A depth map for free:

```python
import matplotlib.pyplot as plt
import numpy as np

vis = pts.in_view
plt.figure(figsize=(14, 7))
plt.scatter(pts.pixels[vis, 0], -pts.pixels[vis, 1], c=pts.depth[vis], s=0.5, cmap="turbo")
plt.colorbar(label="depth (m)"); plt.axis("equal"); plt.title("per-pixel depth from LiDAR")
```

Or open the interactive 3D viewer — the point cloud rendered through the photo's real calibrated camera (drag to orbit, double-click to snap back):
